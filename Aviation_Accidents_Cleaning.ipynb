### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
#inspecting relevant columns first
df[["total.fatal.injuries","total.serious.injuries","total.minor.injuries","total.uninjured"]].isna().sum()
```

```python
# Convert injury counts to numbers, then treat blank counts as zero for this estimate.
injury_columns = ["total.fatal.injuries", "total.serious.injuries", "total.minor.injuries", "total.uninjured"]
for column in injury_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")
df[injury_columns] = df[injury_columns].fillna(0)

# Estimate people on board from the reported injury counts.
df["total_passengers"] = df[injury_columns].sum(axis=1)
df["total_passengers"].describe()
```

```python
# Fatal and serious injuries divided by estimated people on board.
df["serious_injury_fatality_rate"] = (df["total.fatal.injuries"] + df["total.serious.injuries"]) / df["total_passengers"].replace(0, np.nan)
df["serious_injury_fatality_rate"].describe()
```

### Cleaning other columns

I checked engine type, weather, number of engines, purpose of flight, and phase of flight. I removed extra spaces and changed placeholder labels such as `Unknown` and `UNK` to missing values. I left missing values as missing because there is not enough information to fill them accurately.

```python
df["engine.type"].value_counts(dropna=False)
```

```python
# Remove extra spaces and make placeholder values missing
df["engine.type"] = df["engine.type"].str.strip()
df["engine.type"] = df["engine.type"].replace(["Unknown", "UNK", "Unk", "N/A", "NA"], np.nan)
for column in ["weather.condition", "purpose.of.flight", "broad.phase.of.flight"]:
    df[column] = df[column].str.strip()
    df[column] = df[column].replace(["Unknown", "UNK", "Unk", "N/A", "NA"], np.nan)
df["number.of.engines"] = pd.to_numeric(df["number.of.engines"], errors="coerce")
df.loc[df["number.of.engines"] == 0, "number.of.engines"] = np.nan
df[["engine.type", "weather.condition", "number.of.engines", "purpose.of.flight", "broad.phase.of.flight"]].isna().sum()
```

The assignment says to keep columns with more than 20,000 non-missing values. I checked the counts below. I also keep the injury, damage, make/model, and incident-factor columns needed for Part Two, even when they have fewer than 20,000 values.

```python
# Count non-missing values in each column
non_missing_counts = df.notna().sum().sort_values()
non_missing_counts
```

### Save DataFrame to csv
- its generally useful to save data to file/server after its in a sufficiently cleaned or intermediate state
